```python
import os
import json
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from scraper import fetch_website_contents, fetch_website_links
from openai import OpenAI
from rich import print as rich_print
```

#### ONE SHOT PROMPTING
One-shot prompting is a technique where you provide an AI model with exactly one example of the desired behavior, input format, and output format before asking it to complete a new task.

It sits between zero-shot prompting (giving no examples) and few-shot prompting (giving multiple examples).

#### How It Works (The Structure)
A typical one-shot prompt contains three core elements:

__The Instruction:__ A description of the task.

__The Example (The "One Shot"):__ A single pairing of a sample input and its correct output.

__The New Input:__ The actual data you want the AI to process.

#### A Simple Example
Prompt:
You are a translator that converts regular English into pirate slang.

__Example:__

__Input:__ "Hello, where can I find the nearest restaurant?"

__Output:__ "Ahoy, matey! Where be the nearest tavern?"

__Task:__

__Input:__ "Can you help me find my car keys?"

__Output:__

```python
link_system_prompt = """"You are provided with a list of links found on a 
webpage. You are able to decide which of the links would be most relevant 
to include in a brochure about the company, such as links to an About page, 
or a Company page, or Careers/Jobs pages.
You should respond in JSON as in this example:
{
    "relevant_links": [
        {"type": "about page", "url": "https://full.url/goes/here/about"},
        {"type": "careers page", "url": "https://another.full.url/careers"}
    ]
}"""
```

```python
def get_links_attached_to_user_prompt(url):
    user_prompt = f"""Here is the list of links on the website {url} - 
    Please decide which of these are relevant web links for a brochure 
    about the company, respond with the full https URL in JSON format. 
    Do not include Terms of Service, Privacy, email links.
    Links (some might be relative links):"""

    links = fetch_website_links(url)
    user_prompt += "\n".join([f"- {link}" for link in links])
    return user_prompt
```

```python
rich_print(get_links_attached_to_user_prompt("https://edition.cnn.com"))
```

```python
def select_relevant_links(url):
    response = client.chat.completions.create(
        model=model,
        messages=[
            {"role": "system", "content": link_system_prompt},
            {"role": "user", "content": get_links_attached_to_user_prompt(url)}
        ],
        response_format={"type": "json_object"},        
    )
    result = response.choices[0].message.content
    links = json.loads(result)
    rich_print(f"""relevant links: {links}""")
    return links
```

```python
selected_links = select_relevant_links("https://edition.cnn.com")
# rich_print(json.dumps(selected_links, indent=2))

relevant_links = selected_links.get("relevant_links", [])
rich_print(f"""selected links: {relevant_links}""")


if relevant_links:
    md_text = "\n".join(
        f"- [{link.get('type', 'link')}]({url})"
        for link in relevant_links
        for url in ([link["url"]] if "url" in link else link.get("urls", []))
        if url
    )
else:
    fallback_url = selected_links.get("fullUrl") or selected_links.get("url")
    fallback_label = selected_links.get("shortName") or selected_links.get("type", "website")
    md_text = f"- [{fallback_label}]({fallback_url})" if fallback_url else "No relevant links found"


rich_print(f"""md_text: {md_text}""")

display(Markdown(md_text))
```

##### STEP 2: ASSEMBLE ALL THE DETAILS INTO ANOTHER PROMPT TO OLLAMA

```python
def fetch_page_and_all_its_relevant_link(url):
    page_content = fetch_website_contents(url)
    relevant_links = select_relevant_links(url)

    result = f"Page content: {page_content}\n\nRelevant links:\n"
    for link in relevant_links.get("relevant_links", []):
        # Handle both single and multiple URLs
        if "url" in link:
            urls = [link["url"]]
        elif "urls" in link:
            urls = link["urls"]
        else:
            continue  # skip if neither key exists

        for u in urls:
            if not u.startswith(("http://", "https://")):
                u = "https://" + u
            result += f"\n\n### Link: {link.get('type','unknown')} ({u})\n"
            result += fetch_website_contents(u)

    return result

```

```python
rich_print(fetch_page_and_all_its_relevant_link("https://edition.cnn.com"))
```

```python
brochure_system_prompt = """You are an assistant that analyzes the contents of several relevant pages from a company website and creates a short brochure about the company for prospective customers, investors and recruits.
Respond in markdown without code blocks. Include details of company culture, customers and careers/jobs if you have the information."""
```

```python
def get_brochure_user_prompt(company_name, url):
    user_prompt = f"""
You are looking at a company called: {company_name}
Here are the contents of its landing page and other relevant pages;
use this information to build a short brochure of the company in markdown without code blocks.\n\n
"""
    user_prompt += fetch_page_and_all_its_relevant_link(url)
    user_prompt += user_prompt[:7000]
    return user_prompt
```

```python
get_brochure_user_prompt("CNN", "https://edition.cnn.com")
```

```python
def create_brochure(company_name, url):
    response = client.chat.completions.create(
        model=model,
        messages=[
            {"role": "system", "content": brochure_system_prompt},
            {"role": "user", "content": get_brochure_user_prompt(company_name, url)}
        ],
    )
    result = response.choices[0].message.content
    display(Markdown(result))
```

```python
create_brochure("CNN", "https://edition.cnn.com")
```

#### STREAMING

```python
def stream_brochure(company_name, url):
    stream = client.chat.completions.create(
        model=model,
        messages=[
            {"role": "system", "content": brochure_system_prompt},
            {"role": "user", "content": get_brochure_user_prompt(company_name, url)}
        ],
        stream=True
    )
    response = ""
    display_handle = display(Markdown(""), display_id=True)

    for chunk in stream:
        content = chunk.choices[0].delta.content or ""
        response += content
        update_display(Markdown(response), display_id=display_handle.display_id)
```

```python
stream_brochure("CNN", "https://edition.cnn.com")
```
