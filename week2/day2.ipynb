##### What is __ipykernel__ ?
__ipykernel__ stands for __Interactive Python Kernel__. It is the powerful backend engine that executes the code you write inside a Jupyter Notebook cell and sends the results back to your screen.

Think of it as a __translator__:

* __The Frontend (VS Code):__ This is just the UI. It gives you the nice boxes to type in, the run buttons, and the dark theme. It does not know how to run Python.

* __The Backend (Python Engine):__ This is the actual Python interpreter inside your .venv.

* __The Bridge (ipykernel):__ When you press Shift + Enter, VS Code hands the code to ipykernel. ipykernel runs it inside your virtual environment, captures the outputs (or errors), and hands them back to VS Code to display.

```python
import os
from dotenv import load_dotenv
from openai import OpenAI
```

```python
import gradio as gr
```

```python
# Load environment variables from .env file
load_dotenv(override=True)
# Initialize OpenAI client
api_key = os.getenv("OLLAMA_API_KEY")
api_model = os.getenv("OLLAMA_MODEL")
base_url = os.getenv("OLLAMA_BASE_URL")
client = OpenAI(api_key=api_key, base_url=base_url)

# Check if the API key and model are set
if not api_key or not api_model or not base_url:
    raise ValueError("Please set the OLLAMA_API_KEY, OLLAMA_MODEL, and OLLAMA_BASE_URL environment variables in your .env file.")
elif not api_key.startswith("ollama") or not api_model.startswith("llama"):
    raise ValueError("The OLLAMA_API_KEY/OLLAMA_MODEL should start with 'ollama'/'llama'. Please check your .env file.")
else:
    print("Environment variables loaded successfully.")
```

```python
# Let's wrap a call to local OLLAMA 3.2
system_prompt = "You are a helpful assistant that provides concise answers to user queries."

def generate_response(user_prompt):
    response = client.chat.completions.create(
        model=api_model,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt}
        ]
    )
    return response.choices[0].message.content
```

#### STREAMING WITH GRADIO

```python
system_message = "You are a helpful assistant that responds in markdown without code blocks"
```

```python
# Load environment variables from .env file
load_dotenv(override=True)
# Initialize OpenAI client
api_key = os.getenv("GROQ_API_KEY")
api_model = os.getenv("GROQ_MODEL")
base_url = os.getenv("GROQ_BASE_URL")
client = OpenAI(api_key=api_key, base_url=base_url)

# Check if the API key and model are set
if not api_key or not api_model or not base_url:
    raise ValueError("Please set the GROQ_API_KEY, GROQ_MODEL, and GROQ_BASE_URL environment variables in your .env file.")
elif not api_key.startswith("gsk_") or not api_model.startswith("gpt"):
    raise ValueError("The GROQ_API_KEY/GROQ_MODEL should start with 'gsk_'/'gpt'. Please check your .env file.")
else:
    print("Environment variables loaded successfully.")
```

```python
def stream_groq(prompt):
    messages = [
        {"role": "system", "content": system_message},
        {"role": "user", "content": prompt}
      ]
    stream = client.chat.completions.create(
        model=f"openai/{api_model}",
        messages=messages,
        stream=True
    )
    result = ""
    for chunk in stream:
        result += chunk.choices[0].delta.content or ""
        yield result
```

```python
message_input = gr.Textbox(label="Your message:", info="Enter a message for ollama3.2:latest", lines=7)
message_output = gr.Markdown(label="Response:")

view = gr.Interface(
    fn=stream_groq,
    title="Groq Stream Demo", 
    inputs=[message_input], 
    outputs=[message_output], 
    examples=[
        "Explain the Transformer architecture to a layperson",
        "Explain the Transformer architecture to an aspiring AI engineer",
        ], 
    flagging_mode="never"
    )
view.launch(inbrowser=True)
```

#### BUILDIING A COMPANY BROCHURE USING GRADIO AND GROQ

```python
import sys
import os

# Get the path of the parent directory (engineering_llm) and add it to sys.path
sys.path.append(os.path.abspath(os.path.join('..')))
```

```python
from week1.scraper import fetch_website_contents
```

```python
system_message = """
You are an assistant that analyzes the contents of a company website landing page
and creates a short brochure about the company for prospective customers, investors and recruits.
Respond in markdown without code blocks.
"""
```

```python
def stream_brochure(company_name, url, model):
    yield ""
    prompt = f"Please generate a company brochure for {company_name}. Here is their landing page:\n"
    prompt += fetch_website_contents(url)
    if model=="GROQ":
        result = stream_groq(prompt)

    else:
        raise ValueError("Unknown model")
    yield from result
```

```python
name_input = gr.Textbox(label="Company name:")
url_input = gr.Textbox(label="Landing page URL including http:// or https://")
model_selector = gr.Dropdown(["GROQ"], label="Select model", value="GROQ")
message_output = gr.Markdown(label="Response:")

view = gr.Interface(
    fn=stream_brochure,
    title="Brochure Generator", 
    inputs=[name_input, url_input, model_selector], 
    outputs=[message_output], 
    examples=[
            ["Hugging Face", "https://huggingface.co", "GROQ"],
            ["Edward Donner", "https://edwarddonner.com", "GROQ"]
        ], 
    flagging_mode="never"
    )
view.launch(inbrowser=True)
```
