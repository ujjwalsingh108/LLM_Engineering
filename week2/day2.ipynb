## Gradio day!

Today we will build user interfaces using the outrageously popular Gradio library. Gradio allows you to create interactive web applications for your machine learning models with just a few lines of code. Whether you're building a simple demo or a complex application, Gradio makes it easy to share your work with others. Let's dive in and see how we can use Gradio to create an interface for our machine learning model!

```python
import os
from dotenv import load_dotenv
from openai import OpenAI
```

```python
import gradio as gr
```

```python
# Load environment variables from .env file
load_dotenv(override=True)
# Initialize OpenAI client
api_key = os.getenv("OLLAMA_API_KEY")
api_model = os.getenv("OLLAMA_MODEL")
base_url = os.getenv("OLLAMA_BASE_URL")
client = OpenAI(api_key=api_key, base_url=base_url)

# Check if the API key and model are set
if not api_key or not api_model or not base_url:
    raise ValueError("Please set the OLLAMA_API_KEY, OLLAMA_MODEL, and OLLAMA_BASE_URL environment variables in your .env file.")
elif not api_key.startswith("ollama") or not api_model.startswith("llama"):
    raise ValueError("The OLLAMA_API_KEY/OLLAMA_MODEL should start with 'ollama'/'llama'. Please check your .env file.")
else:
    print("Environment variables loaded successfully.")
```

```python
# Let's wrap a call to local OLLAMA 3.2
system_prompt = "You are a helpful assistant that provides concise answers to user queries."

def generate_response(user_prompt):
    response = client.chat.completions.create(
        model=api_model,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt}
        ]
    )
    return response.choices[0].message.content
```

```python
# generate_response("What is the capital of France?")
generate_response("What is today's date?")
```

## User Interface time!

```python
# Here is a simple function to create a Gradio interface for our chatbot
def shout(text):
    print(f"Shout has been called with an input: {text}")
    return text.upper()
```

```python
shout("hello world")
```

```python
gr.Interface(fn=shout, inputs="textbox", outputs="textbox", flagging_mode="never").launch()
```

## Gradio share tool

```python
gr.Interface(fn=shout, inputs="text", outputs="text", flagging_mode="never").launch(share=True)
```

```python
# Adding inbrowser=True opens up a new browser window automatically

gr.Interface(fn=shout, inputs="textbox", outputs="textbox", flagging_mode="never").launch(inbrowser=True, auth=("admin", "password"))
```

```python
# Define this variable and then pass js=force_dark_mode when launching the Gradio interface to enable dark mode
force_dark_mode = """
    function refresh() {
        const url = new URL(window.location);
        if (url.searchParams.get('__theme') !== 'dark') {
            url.searchParams.set('__theme', 'dark');
            window.location.href = url.href;
        }
    }
"""

gr.Interface(fn=shout, inputs="textbox", outputs="textbox", flagging_mode="never", js=force_dark_mode).launch(inbrowser=True)
```

```python
# Adding a little more:

msg_input = gr.Textbox(label="Enter your message here", info="Type something and I'll shout it back at you!", lines=2, max_lines=5)
msg_output = gr.Textbox(label="Shouted message will appear here", info="This will show the uppercase version of your input.", lines=2, max_lines=5)

view = gr.Interface(fn=shout, inputs=msg_input, outputs=msg_output, flagging_mode="never", examples=[["Hello, Gradio!"], ["How are you doing today?"]], title="Shout Bot", description="A simple app that shouts back whatever you type in uppercase letters.")
view.launch(inbrowser=True)
```

```python
msg_input = gr.Textbox(label="Enter your message here for Ollama 3.2", info="Type something and I'll shout it back at you!", lines=2, max_lines=5)
msg_output = gr.Textbox(label="Shouted message will appear here", info="This will show the uppercase version of your input.", lines=2, max_lines=5)

view = gr.Interface(fn=generate_response, inputs=msg_input, outputs=msg_output, flagging_mode="never", examples=[["Hello, Gradio!"], ["How are you doing today?"]], title="Shout Bot Ollama", description="A simple app that shouts back whatever you type in uppercase letters.")
view.launch(inbrowser=True)
```
